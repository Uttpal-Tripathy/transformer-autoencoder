```python
import sys, os, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))

try:
    from dotenv import load_dotenv
    load_dotenv(pathlib.Path.cwd().parent / ".env")
except ImportError:
    pass  # optional; set the env vars another way if you don't use python-dotenv

assert os.environ.get("OPENAI_API_KEY"), "Set OPENAI_API_KEY in your environment or .env"
assert os.environ.get("ANTHROPIC_API_KEY"), "Set ANTHROPIC_API_KEY in your environment or .env"
```
