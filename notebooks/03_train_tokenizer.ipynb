# 3. Train the tokenizer

Trains a byte-level BPE tokenizer to `vocab_size=32000` on `data/corpus.txt`, matching the diagram's "Tokenizer / Vocabulary 32,000" box. Special tokens `<pad> <bos> <eos> <unk> <mask>` are pinned to ids 0-4 to match `model/config.py`.

```python
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))

from tokenizer.train_tokenizer import train

corpus_path = "../data/corpus.txt"
out_dir = "../tokenizer/vocab"
assert pathlib.Path(corpus_path).exists(), "Run 02_generate_corpus.ipynb first"

train(corpus_path, out_dir, vocab_size=32000)
```

Sanity-check the trained tokenizer by round-tripping a sentence.

```python
from tokenizers import ByteLevelBPETokenizer

tok = ByteLevelBPETokenizer(f"{out_dir}/vocab.json", f"{out_dir}/merges.txt")
sample = "The transformer autoencoder reconstructs corrupted text."
enc = tok.encode(sample)
print("token ids:", enc.ids)
print("decoded:  ", tok.decode(enc.ids))
```
